```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("empirical_results.csv")
df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.lineplot(data=df, x='Files', y='Time(seconds)', hue='Algorithm', marker='o')
plt.title("Algorithm Execution Time (Linear Scale)", fontsize=14, fontweight="bold")
plt.xlabel("Number of Files (N)", fontsize=12)
plt.ylabel("Execution Time (Seconds)", fontsize=12)
plt.tight_layout()
plt.savefig('benchmark_plot.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.lineplot(data=df, x="Files", y="Time(seconds)", hue="Algorithm", marker="o",)

# Transform the Y-axis to a logarithmic scale
plt.yscale("log")

plt.title("Algorithm Execution Time (Logarithmic Scale)", fontsize=14, fontweight="bold")
plt.xlabel("Number of Files (N)", fontsize=12)
plt.ylabel("Execution Time (Seconds - Log Scale)", fontsize=12)
plt.tight_layout()
plt.savefig('benchmark_plot_log.png')
plt.show()
```
